### Part (a) -- 1pt

Write a function `sum_of_cubes` that computes the sum of cubes up to `n`. If the input to `sum_of_cubes` invalid (e.g. negative or non-integer `n`), the function should print out `"Invalid input"` and return `-1`.

```python
def sum_of_cubes(n):
    """Return the sum (1^3 + 2^3 + 3^3 + ... + n^3)
    
    Precondition: n > 0, type(n) == int
    
    >>> sum_of_cubes(3)
    36
    >>> sum_of_cubes(1)
    1
    """
    # Check for invalid input
    if type(n) is not int or n <= 0:
        print("Invalid input")
        return -1

    # Calculate the sum of cubes
    total = 0
    for i in range(1, n + 1):
        total += i ** 3
    return total

# Test cases
print(sum_of_cubes(3))
print(sum_of_cubes(1))
sum_of_cubes(-1)  # Invalid input
sum_of_cubes(2.5)  # Invalid input
print(sum_of_cubes(5))
sum_of_cubes("three")  # Invalid input
sum_of_cubes("3")  # Invalid input
```

Write a function `word_lengths` that takes a sentence (string), computes the length of each word in that sentence, and returns the length of each word in a list. You can

Hint: recall the `str.split` function in Python.

```python
def word_lengths(sentence):
    """Return a list containing the length of each word in
    sentence.
    
    >>> word_lengths("welcome to APS360!")
    [7, 2, 7]
    >>> word_lengths("machine learning is so cool")
    [7, 8, 2, 2, 4]
    """
    # Split the sentence into words
    words = sentence.split()
    
    # Calculate the length of each word
    lengths = [len(word) for word in words]
    
    return lengths

# Test cases
print(word_lengths("welcome to APS360!"))
print(word_lengths("machine learning is so cool"))
print(word_lengths("This year is going to be\n great!"))
```

### Part (c) -- 1pt

Write a function `all_same_length` that takes a sentence (string),
and checks whether every word in the string is the same length.

```python
def all_same_length(sentence):
    """Return True if every word in sentence has the same
    length, and False otherwise.
    
    >>> all_same_length("all same length")
    False
    >>> all_same_length("hello world")
    True
    """
    # Get the lengths of the words in the sentence
    array = word_lengths(sentence)
    # Check if all lengths are the same
    return all(length == array[0] for length in array)

# Test cases
print(all_same_length("all same length"))
print(all_same_length("hello world"))
print(all_same_length("test test quiz"))
print(all_same_length("one two three"))
```

## Part 5. Training an ANN [10 pt]

The sample code provided below is a 2-layer ANN trained on the MNIST dataset to identify digits less than 3 or greater than and equal to 3. Modify the code by changing any of the following and observe how the accuracy and error are affected:

- number of training iterations
- number of hidden units
- numbers of layers
- types of activation functions
- learning rate

Please select at least three different options from the list above. For each option, please select two to three different parameters and provide a table. 


```python
import torch
import torch.nn as nn
import torch.nn.functional as F
from torchvision import datasets, transforms
import matplotlib.pyplot as plt # for plotting
import torch.optim as optim

torch.manual_seed(1) # set the random seed

# define a 2-layer artificial neural network
class Pigeon(nn.Module):
    def __init__(self):
        super(Pigeon, self).__init__()
        self.layer1 = nn.Linear(28 * 28, 30)
        self.layer2 = nn.Linear(30, 1)
    def forward(self, img):
        flattened = img.view(-1, 28 * 28)
        activation1 = self.layer1(flattened)
        activation1 = F.relu(activation1)
        activation2 = self.layer2(activation1)
        return activation2

pigeon = Pigeon()

# load the data
mnist_data = datasets.MNIST('data', train=True, download=True)
mnist_data = list(mnist_data)
mnist_train = mnist_data[:1000]
mnist_val   = mnist_data[1000:2000]
img_to_tensor = transforms.ToTensor()
      
    
# simplified training code to train `pigeon` on the "small digit recognition" task
criterion = nn.BCEWithLogitsLoss()
optimizer = optim.SGD(pigeon.parameters(), lr=0.005, momentum=0.9)

# number of iterations 
num_iterations = 1

# training loop
for i in range(num_iterations):
    for (image, label) in mnist_train:
        # actual ground truth: is the digit less than 3?
        actual = torch.tensor(label < 3).reshape([1,1]).type(torch.FloatTensor)
        # pigeon prediction
        out = pigeon(img_to_tensor(image)) # step 1-2
        # update the parameters based on the loss
        loss = criterion(out, actual)      # step 3
        loss.backward()                    # step 4 (compute the updates for each parameter)
        optimizer.step()                   # step 4 (make the updates for each parameter)
        optimizer.zero_grad()              # a clean up step for PyTorch

# computing the error and accuracy on the training set
error = 0
for (image, label) in mnist_train:
    prob = torch.sigmoid(pigeon(img_to_tensor(image)))
    if (prob < 0.5 and label < 3) or (prob >= 0.5 and label >= 3):
        error += 1
print("Training Error Rate:", error/len(mnist_train))
print("Training Accuracy:", 1 - error/len(mnist_train))


# computing the error and accuracy on a test set
error = 0
for (image, label) in mnist_val:
    prob = torch.sigmoid(pigeon(img_to_tensor(image)))
    if (prob < 0.5 and label < 3) or (prob >= 0.5 and label >= 3):
        error += 1
print("Test Error Rate:", error/len(mnist_val))
print("Test Accuracy:", 1 - error/len(mnist_val))
```

## Part 5 Experiment Summary

The baseline model uses 1,000 training images, 30 hidden units, ReLU activation, a learning rate of `0.005`, and one training pass through the data.

### 1. Training Iterations

| Training iterations | Training accuracy | Testing accuracy |
|---:|---:|---:|
| 1 | 96.4% | 92.1% |
| 2 | 98.4% | 94.3% |
| 3 | 98.6% | 93.1% |

Additional iterations give the optimizer more opportunities to update the weights using gradient descent, reducing the training loss and improving training accuracy. However, after two iterations, the model begins fitting the training examples too closely. This mild overfitting explains why training accuracy continued to increase while testing accuracy decreased slightly after three iterations.

### 2. Number of Hidden Units

| Hidden units | Training accuracy | Testing accuracy |
|---:|---:|---:|
| 15 | 96.2% | 91.3% |
| 30 | 96.4% | 92.1% |
| 60 | 96.9% | 92.0% |

Increasing the number of hidden units increases the model's capacity because it provides more learnable weights and allows the network to represent more complex patterns. This slightly improved training accuracy. However, the testing accuracy changed very little, suggesting that the task and limited training set did not require the additional capacity. Extra units can also increase the risk of learning noise rather than generalizable features.

### 3. Number of Layers

| Network architecture | Training accuracy | Testing accuracy |
|---|---:|---:|
| One hidden layer | 96.4% | 92.1% |
| Two hidden layers | 96.1% | 91.3% |
| Three hidden layers | 95.9% | 90.7% |

Deeper networks can learn hierarchical representations, where earlier layers detect simpler patterns and later layers combine them into more complex features. However, adding layers also introduces more parameters and makes optimization more difficult. During backpropagation, gradients must pass through each layer, so they can become very small or unstable. With only 1,000 training examples, the deeper networks were harder to optimize and produced lower accuracy.
### 4. Activation Function

| Activation function | Training accuracy | Testing accuracy |
|---|---:|---:|
| ReLU | 96.4% | 92.1% |
| Sigmoid | 92.7% | 88.3% |
| Tanh | 96.0% | 90.6% |

ReLU outputs zero for negative inputs and keeps positive inputs unchanged. This produces useful gradients for active neurons and generally makes gradient-based learning easier.

Sigmoid and Tanh compress their inputs into limited ranges. For large positive or negative inputs, their gradients become very small, causing the vanishing-gradient problem. As a result, earlier layers receive weaker updates and learn more slowly. ReLU therefore achieved the best performance in this experiment.

### 5. Learning Rate

| Learning rate | Training accuracy | Testing accuracy |
|---:|---:|---:|
| 0.001 | 92.2% | 88.7% |
| 0.005 | 96.4% | 92.1% |
| 0.01 | 96.1% | 91.8% |

The learning rate controls the size of each gradient-descent update. With a learning rate of `0.001`, the updates were very small, so the model learned slowly and did not reach as effective a set of weights after one training pass.

The rate of `0.005` provided a good balance between learning speed and stability. Increasing the rate to `0.01` made the updates more drastic. These larger updates can overshoot low points in the loss function instead of settling near them, which explains the slight decrease in training and testing accuracy.
